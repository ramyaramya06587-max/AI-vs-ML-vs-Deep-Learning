## 1. Artificial Intelligence (AI)

Artificial Intelligence (AI) is a branch of computer science that focuses on creating machines capable of performing tasks that typically require human intelligence.

### Real-World Examples
- Virtual assistants
- Intelligent robots
- Rule-based customer support systems
- AI planning systems

### Applications
- Healthcare
- Education
- Robotics
- Customer support

## 2. Machine Learning (ML)

Machine Learning (ML) is a subset of Artificial Intelligence that enables computers to learn patterns from data and make predictions or decisions without being explicitly programmed for every situation.

### Real-World Examples
- Email spam detection
- Product recommendation systems
- Fraud detection
- House price prediction

### Applications
- Banking and finance
- E-commerce
- Healthcare
- Predictive analytics

## 3. Deep Learning (DL)

Deep Learning (DL) is a subset of Machine Learning that uses artificial neural networks with multiple layers to learn complex patterns from large amounts of data.

### Real-World Examples
- Facial recognition
- Speech recognition
- Image classification
- Object detection

### Applications
- Self-driving vehicles
- Medical image analysis
- Voice assistants
- Computer vision

## 4. Comparison Between AI, ML, and Deep Learning

| Feature | Artificial Intelligence (AI) | Machine Learning (ML) | Deep Learning (DL) |
|---|---|---|---|
| Definition | A broad field that creates intelligent systems | A subset of AI that learns from data | A subset of ML that uses neural networks with multiple layers |
| Main Goal | Perform tasks requiring intelligence | Learn patterns and make predictions | Learn complex patterns from data |
| Techniques | Rules, search, planning, and ML | Statistical methods and learning algorithms | Multi-layer neural networks |
| Data Requirement | Depends on the approach | Usually requires training data | Often benefits from large datasets |
| Human Involvement | Depends on the system | May require feature selection and preparation | Can automatically learn useful features |
| Example | Rule-based virtual assistant | Email spam detection | Image recognition |
| Applications | Robotics and planning | Recommendations and fraud detection | Speech and image recognition |

## 5. Relationship Between AI, ML, and Deep Learning

Artificial Intelligence (AI) is the broadest field, Machine Learning (ML) is a subset of AI, and Deep Learning (DL) is a subset of ML.

### Relationship
- AI includes different techniques for creating intelligent systems.
- ML is a part of AI that learns patterns from data.
- DL is a part of ML that uses multi-layer neural networks.

### Example

An AI system can use predefined rules to answer questions. A Machine Learning system can learn to identify spam emails from examples. A Deep Learning system can use neural networks to recognize objects in images.

### Key Point

**Deep Learning is a subset of Machine Learning, and Machine Learning is a subset of Artificial Intelligence.**

Not every AI system uses ML, and not every ML system uses DL.
